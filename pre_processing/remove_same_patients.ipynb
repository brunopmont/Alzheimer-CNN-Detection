```python
import os
import shutil
```

```python
dir_base = "/mnt/c/Users/Bruno/Desktop/IANS/ADNI"

# dir_ad = f"{dir_base}/ad"
# dir_mci = f"{dir_base}/mci"
# dir_cn = f"{dir_base}/cn"

input_dir = os.path.join(dir_base, "ADNI_3_4")
output_dir = os.path.join(dir_base, "ADNI_3_4_RAW")

labels = ['cn', 'mci', 'ad']

for label in labels:
    output_dir_create = os.path.join(output_dir, label)
    os.makedirs(output_dir_create, exist_ok=True)
```

```python
for label in labels:
    subjects = os.listdir(os.path.join(input_dir, label))
    for sub in subjects:
        names = os.listdir(os.path.join(input_dir, label, sub))
        datas = os.listdir(os.path.join(input_dir, label, sub, names[0]))
        file = os.listdir(os.path.join(input_dir, label, sub, names[0], datas[0]))
        file_path = os.path.join(input_dir, label, sub, names[0], datas[0], file[0])
        output_path = os.path.join(output_dir, label, file[0])
        print(file_path)
        shutil.copytree(file_path, output_path)
```
